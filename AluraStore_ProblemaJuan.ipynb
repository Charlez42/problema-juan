# 2. Ventas por categoría

## **Cantidad de productos vendidos y más populares en Tienda**

```python
ventas_por_categoria = tienda.groupby(['Lugar de Compra', 'Categoría del Producto']).size().reset_index(name='Cantidad Vendida')
ventas_por_categoria
```

```python
tiendas = ventas_por_categoria['Lugar de Compra'].unique()
tiendas
```

```python
for lugar in tiendas:
    datos_tienda = ventas_por_categoria[ventas_por_categoria['Lugar de Compra'] == lugar]

    datos_ordenados = datos_tienda.sort_values(by='Cantidad Vendida', ascending=False)

    print(f"Categorías más populares en {lugar}:")
    print(datos_ordenados.to_string(index=False))
```
