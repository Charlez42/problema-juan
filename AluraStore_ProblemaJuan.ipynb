## VISUALIZACIÓN DE GRÁFICOS: PRODUCTOS MÁS VENDIDOS Y MENOS VENDIDOS

```python
urls = {"Tienda": url,"Tienda 2": url2,"Tienda 3": url3,"Tienda 4": url4}

def graficar_barras(nombre_tienda, url):
    df = pd.read_csv(url)

    # Agrupar ventas por producto
    ventas = df.groupby("Producto").size().reset_index(name="Cantidad Vendida")

    # Ordenar de mayor a menor
    ventas = ventas.sort_values("Cantidad Vendida", ascending=True)

    # Crear gráfico
    plt.figure(figsize=(10, 8))
    plt.barh(ventas["Producto"], ventas["Cantidad Vendida"], color="skyblue")

    plt.title(f"Productos más vendidos – {nombre_tienda}")
    plt.xlabel("Cantidad vendida")
    plt.ylabel("Producto")

    plt.tight_layout()
    plt.show()

# Generar un gráfico por tienda
for nombre, url in urls.items():
    graficar_barras(nombre, url)
```

## VISUALIZACIÓN DE GRÁFICOS: ENVÍO PROMEDIO POR TIENDA

```python
# URLs de las tiendas
urls = {
    "Tienda": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "Tienda 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "Tienda 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "Tienda 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"
}

# Calcular promedios
promedios = {}

for nombre, url in urls.items():
    dfprom = pd.read_csv(url)
    promedios[nombre] = dfprom.groupby("Lugar de Compra")["Costo de envío"].mean()

# Convertir a DataFrame para graficar fácilmente
dfprom_promedios = pd.DataFrame(promedios)

# Crear gráfico
plt.figure(figsize=(12, 6))
dfprom_promedios.plot(kind="bar", figsize=(12, 6))

plt.title("Costo de envío promedio por lugar de compra y tienda")
plt.xlabel("Lugar de compra")
plt.ylabel("Costo de pnvío promedio (en pesos colombianos)")
plt.xticks(rotation=45)
plt.legend(title="Tiendas")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
